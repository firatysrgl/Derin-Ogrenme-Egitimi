```python
def divideNumber(number):
    return number /2
```

```python
divideNumber (20)
```

```python
myList= [3,5,7,10,20,30]
```

```python
myResultList = []
for num in myList :
    myResultList.append(divideNumber(num))
    print (myResultList)
```

# Map

```python
help(map)
```

```python
def controlString(string):
    return "fırat" in string 
```

```python
controlString ("fırat Yaşaroğlu")
```

```python
myStringList = ["fırat", "yunus", "Yaşaroğlu", "firat", "fırat yunus Yaşaroğlu"]
```

```python
list(map(controlString,myStringList))
```

# Filter

```python
list (filter(controlString,myStringList))
```
