# Polymorphism

```python
class Banana():
    def __init__(self,name):
        self.name=name
    def info (self):
        return f"100 calories {self.name}"
class Apple():
    def __init__(self,name):
        self.name=name
    def info (self):
        return f"150 calories {self.name}"
```

```python
banana=Banana("Banana")
```

```python
apple=Apple("Apple")
```

```python
banana.info()
```

```python
apple.info()
```

```python
fruitList = [apple,banana]
```

```python
for fruit in fruitList:
    print(fruit.info())
```

# Encapsulation [Dışarıdan belirlenen değişkene müdahale etmeyi engeller.]

```python
class Phone():
    
    def __init__(self,name,price):
        self.name=name
        self.__price=price

    def info(self):
        print(f"{self.name} price is: {self.__price}")
        
        # !!! 2 tane "_" eklersek değişkene dışarıdan müdahaleye kapalı olur. 

    def changePrice(self,price):
        self.__price = price
```

```python
iphone=Phone("İphone 14",500)
```

```python
iphone.info()
```

```python
iphone.price = 400 ## dışardan değişken değerini değiştirmeye çalışsakta değişmiyor. sadece class içinden değiştirebiliriz.
```

```python
iphone.info()
```

```python
iphone.changePrice(300)
```

```python
iphone.info()
```

# Abstraction

```python
from abc import ABC, abstractmethod
```

```python
class Car (ABC):

    @abstractmethod
    def maxSpeed():
        pass
```

```python
class Tesla (Car):

    def maxSpeed(self):
        print ("200km/h")
```

```python
tesla =Tesla()
```

```python
tesla.maxSpeed()
```

```python
class Mercedes (Car):

    def maxSpeed (self):
        print ("250Km/h")
```

```python
mercedes=Mercedes()
```

```python
mercedes.maxSpeed()
```
