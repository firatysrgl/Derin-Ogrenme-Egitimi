```python
class Musician ():
    def __init__(self,name):
        self.name=name 
        print ("Musician Class")
    def test1(self):
        print ("test1")
    def test2(self):
        print  ("test2")
    
```

```python
BBS = Musician("BBS")
```

```python
BBS.name
```

```python
BBS.test1()
```

```python
BBS.test2()
```

```python
class MusicianPlus(Musician):
    def __init__(self,name):
        Musician.__init__(self,name)
        print("Musician Plus")
    def test3(self):
        print("test3")
        #OVERRİDE
    def test1(self):
        print("TEST1","TEST1","TEST1")
```

```python
atlas = MusicianPlus("atlas")
```
