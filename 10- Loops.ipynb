```python
my_list = [10,20,30,40,50,60,70]
```

```python
print ("for loop started")
for num in my_list:
    print (num)
print ("for loop ended")
```

```python
40 in my_list 
```

```python
for number in my_list :
    print (number)
    if number ==40:
        print ("yes")
```

```python
for number in my_list :
    print (number)
    if number ==40:
        print ("yes")     # döngüyü bitirir 
        break 
```

```python
for num in my_list:
    print (num)
    if num == 40:
        print ("yesss")   # deavm etmeyi sağlar 
        continue
```

```python
for num  in my_list:
    pass                     # Pas geçmek için kullanılır 
```
